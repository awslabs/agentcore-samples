## Step 2: Grant Lake Formation Admin Permissions (One-time Setup)

Before proceeding, your AWS role needs Lake Formation administrator permissions to create the federated catalog.

**Option 1: AWS Console (Recommended)**
1. Go to [AWS Lake Formation console](https://console.aws.amazon.com/lakeformation/)
2. Navigate to "Administrative roles and tasks" → "Data lake administrators"
3. Click "Choose administrators"
4. Add your IAM role ARN (check the output from cell 1 for your role)
5. Click "Save"

**Option 2: run the cell below** — an idempotent, non-destructive helper that adds *your current role* to the data lake admins by **union**: it reads the existing admins and appends your role only if it is missing.

> ⚠️ **Do not use a raw `aws lakeformation put-data-lake-settings` CLI call here.** That API **replaces** the entire `DataLakeAdmins` list rather than appending to it, so it would silently **strip every existing administrator** — including ones you did not create (for example a shared account's SageMaker or analytics admin). The cell below performs the same grant safely by reading the current list first and writing back existing + yours.

**After granting permissions (either option), continue to the next cell.**

````python
# Option 2 (executable): idempotently add THIS role to the Lake Formation data
# lake admins by UNION -- read the current admins, append only if missing, and
# never replace the list.
#
# Why not the CLI: `aws lakeformation put-data-lake-settings` OVERWRITES the whole
# DataLakeAdmins list. On a shared account that silently removes other teams'
# admins. This cell reads first, appends only what is missing, and writes back
# existing + yours -- so it is safe to re-run and safe on a shared account.
import boto3

lf_client = boto3.client("lakeformation", region_name=region)
sts_client = boto3.client("sts", region_name=region)

identity = sts_client.get_caller_identity()
caller_arn = identity["Arn"]

# Data lake admins are IAM role/user ARNs. Convert an assumed-role SESSION arn
# (arn:aws:sts::<acct>:assumed-role/<Role>/<session>) to its IAM ROLE arn.
if ":assumed-role/" in caller_arn:
    role_name = caller_arn.split(":assumed-role/")[1].split("/")[0]
    principal = f"arn:aws:iam::{account_id}:role/{role_name}"
else:
    principal = caller_arn

settings = lf_client.get_data_lake_settings()["DataLakeSettings"]
admins = settings.get("DataLakeAdmins", [])
print("Data lake admins (before):", [a["DataLakePrincipalIdentifier"] for a in admins])

if principal in [a["DataLakePrincipalIdentifier"] for a in admins]:
    print(f"✅ Already a data lake admin: {principal}")
else:
    # UNION: everything that was already there, plus this role.
    admins.append({"DataLakePrincipalIdentifier": principal})
    settings["DataLakeAdmins"] = admins
    # get_data_lake_settings returns ReadOnlyAdmins, but put rejects the pair on
    # some accounts -- drop it so the write succeeds.
    settings.pop("ReadOnlyAdmins", None)
    lf_client.put_data_lake_settings(DataLakeSettings=settings)
    after = lf_client.get_data_lake_settings()["DataLakeSettings"].get("DataLakeAdmins", [])
    print(f"✅ Added {principal}")
    print("Data lake admins (after): ", [a["DataLakePrincipalIdentifier"] for a in after])
````

## Step 5: Configure Lake Formation Permissions

**Column filtering** removes sensitive *columns* (e.g. PII such as date of birth) from a role's view of the table, and **table grants** decide which roles may read a table at all. These Lake Formation grants are what enforce per-persona *column* access downstream: the REQUEST interceptor assumes a tenant role, and Lake Formation applies that role's column filters and grants to every Athena query.

Note the word: **filtering**, not masking. Nothing is redacted or blanked — the column is simply **absent** from the table as that role sees it. Lake Formation filters the table *definition* per role, so a `glue:GetTable` under a tenant role returns only the granted columns, and `SELECT *` expands to exactly that set: 15 columns for policyholders, 20 for adjusters (all but `policyholder_dob`), 21 for administrators. That is why the claims tools select `*` rather than naming columns — the projection is deliberately implicit, so the tool inherits the grant instead of keeping a second copy of it that could drift.

**Where the guarantee actually lives.** Column filtering is not enforced inside the storage layer. AWS's Lake Formation permissions reference is explicit that "it is the responsibility of the integrated analytics service to apply the column filtering when processing a query" — here that service is Athena. So these grants are authoritative for queries that go through Athena, which is every query this sample makes, but they are not a property of the underlying S3 Tables data itself. Worth knowing before you rely on them as the only control.

Per-user **row** scope is a different mechanism in this sample — the claims tools bind the interceptor-propagated caller identity into a `WHERE user_id = ?` predicate. Lake Formation data-cell filters are **not** used, and the reason is structural rather than a shortcut: a filter's row expression takes **constants only** and permits no PartiQL functions, so it cannot reference the caller. Per-user scope would mean one filter per user. Full reasoning and AWS references in `deployment/3-s3tables-setup/setup_lakeformation_permissions.py`.

Set up Lake Formation permissions for tenant roles (policyholders, adjusters, administrators). This must be done BEFORE loading data so the admin role has INSERT permissions.

⏱️ Lake Formation grants can take a few seconds to propagate before queries reflect them.

## Step 7: Validate Deployment

Verify that the database and tables were created successfully.

````python
print("🔍 Validating S3 Tables deployment...\n")

# Use session from cell 1 to create AWS clients
s3tables_client = session.client("s3tables", region_name=region)
glue_client = session.client("glue", region_name=region)

# Get catalog name and database name from SSM
try:
    catalog_response = ssm_client.get_parameter(Name="/app/lakehouse-agent/catalog-name")
    catalog_name = catalog_response["Parameter"]["Value"]
    print(f"📚 Using catalog: {catalog_name}")
except:
    catalog_name = "s3tablescatalog"
    print(f"⚠️  Using default catalog: {catalog_name}")

try:
    db_response = ssm_client.get_parameter(Name="/app/lakehouse-agent/database-name")
    database_name = db_response["Parameter"]["Value"]
except:
    database_name = "lakehouse_data"
print(f"📚 Using database: {database_name}")

# For S3 Tables federated catalog, CatalogId = account_id:catalog_name
catalog_id = f"{account_id}:{catalog_name}"
print(f"📚 Catalog ID: {catalog_id}")

# Check database exists
try:
    response = glue_client.get_database(CatalogId=catalog_id, Name=database_name)
    print(f"✅ Database '{database_name}' exists in catalog '{catalog_name}'")
except glue_client.exceptions.EntityNotFoundException:
    print(f"❌ Database '{database_name}' not found in catalog '{catalog_name}'")

# Check tables exist
try:
    response = glue_client.get_tables(CatalogId=catalog_id, DatabaseName=database_name)
    tables = response["TableList"]

    print(f"\n📋 Tables in {catalog_name}.{database_name}:")
    for table in tables:
        table_name = table["Name"]
        column_count = len(table["StorageDescriptor"]["Columns"])
        print(f"   • {table_name} ({column_count} columns)")

    if len(tables) >= 2:
        print("\n✅ All tables created successfully")
    else:
        print(f"\n⚠️  Expected 2 tables, found {len(tables)}")

except Exception as e:
    print(f"❌ Error checking tables: {e}")
````
